# Actividad 4.3: Transformación de Datos

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, OneHotEncoder

df = pd.DataFrame({
    'precio': [1200, 1500, 3000, 4500],
    'habitaciones': [1, 2, 3, 4],
    'tipo_alojamiento': ['Privado', 'Compartido', 'Privado', 'Hotel']
})
```

## Aplicando Transformaciones

```python
# Normalización (Min-Max)
scaler_mm = MinMaxScaler()
df['precio_minmax'] = scaler_mm.fit_transform(df[['precio']])

# Estandarización (Z-score)
scaler_std = StandardScaler()
df['precio_zscore'] = scaler_std.fit_transform(df[['precio']])

# Codificación de Variables Categóricas (One-Hot Encoding)
encoder = OneHotEncoder(sparse_output=False)
encoded_tipo = encoder.fit_transform(df[['tipo_alojamiento']])
df_encoded = pd.DataFrame(encoded_tipo, columns=encoder.get_feature_names_out())

# Creación de Variables Derivadas
df['precio_por_hab'] = df['precio'] / df['habitaciones']

print("--- DataFrame Final con Transformaciones ---")
print(df.join(df_encoded).drop(columns=['tipo_alojamiento']))
```

### ¿Cúando usar cada uno?

1. Normalización (Min-Max)
    * ¿Cuándo usarla? Cuando no tenemos outliers y cuando los algoritmos no asumen una distribución específica

2. Estandarización (Z-score)
    * ¿Cuándo usarla? Se usa para la mayoría de los algoritmos de Machine Learning (Regresiones, SVM, PCA) porque ayuda a que el modelo converja mucho más rápido y no se pierda en escalas diferentes.

3. Codificación Categórica (One-Hot Encoding)
    * ¿Cuándo usarla? Cuando las categorías no tienen un orden lógico

4. Variables Derivadas (Feature Engineering)
    * ¿Cuándo usarla? Siempre que se pueda inyectar "conocimiento humano" al modelo.
