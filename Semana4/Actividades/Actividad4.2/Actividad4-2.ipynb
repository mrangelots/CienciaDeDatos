# Actividad 4.2: Imputación de Datos

```python
import numpy as np 
import pandas as pd

df = pd.DataFrame({
    'precio': [1200, np.nan, 1500, 1200, 3000, np.nan],
    'habitaciones': [1, 2, np.nan, 1, 3, 2],
    'barrio': ['Centro', 'Sur', 'Centro', np.nan, 'Norte', 'Norte']
})

print(f"- - - DataFrame con Nulos - - -\n{df}")
```

## Aplicando técnicas

```python
# Imputación con Media
df_media = df.copy()
df_media['precio'] = df_media['precio'].fillna(df_media['precio'].mean())

# Imputación con Mediana
df_mediana = df.copy()
df_mediana['precio'] = df_mediana['precio'].fillna(df_mediana['precio'].median())

# Imputación con Moda
df_moda = df.copy()
df_moda['barrio'] = df_moda['barrio'].fillna(df_moda['barrio'].mode()[0])

# Imputación hacia adelante y atrás (Foward Fill y Backward Fill)
df_ffill = df.ffill()
df_bfill = df.bfill()
```

```python
print("\n--- Comparativa de Resultados (Columna Precio) ---")
comparativa = pd.DataFrame({
    'Original': df['precio'],
    'Media': df_media['precio'],
    'Mediana': df_mediana['precio'],
    'Forward_Fill': df_ffill['precio']
})
print(comparativa)
```

### ¿Cúando usar cada metodo?

| Metodo        | Cuándo Usarlo                                 | Justificación                                         |
|---------------|-----------------------------------------------|-------------------------------------------------------|
| Media         | Datos númericos sin valores extremos          | Mantiene el promedio general del dataset              |
| Mediana       | Datos numéricos con outliers                  | Es más robusta                                        |
| Moda          | Variables categóricas                         | Cuando necesitamos promediar                          |
| Foward Fill   | Datos ordenados en el tiempo                  | Se asume que el precio de hoy es igual al de ayer     |
| Backward Fill | Datos donde el valor futuro es buen predictor.| útil si el dato es parte de un proceso que ya sabemos |
