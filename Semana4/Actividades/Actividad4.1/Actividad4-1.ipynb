# Actividad 4.1: Identificación de Valores Faltantes

```python
import pandas as pd
import numpy as np

# Creamos nuestro DataFrame 
df = pd.DataFrame({
    'precio': [1200, np.nan, 1500, 1200, 3000, np.nan],
    'habitación': [1, 2, np.nan, 1, 3, 2],
    'barrio': ['Centro', 'Sur', 'Centro', np.nan, 'Norte', 'Norte']
})

# Contamos los valores nulos por columna
print(f"Nulos por columna:\n{df.isnull().sum()}")
```

## Tecnicas de manejo

```python
# Eliminación de filas con nulos 
df_borrado = df.dropna()

# Imputación con mediana para lo valores numéricos 
df_mediana = df.copy()
df_mediana['precio'] = df_mediana['precio'].fillna(df_mediana['precio'].median())

# Imputación con moda para valores categóricos
df_moda = df.copy()
df_moda['barrio'] = df_moda['barrio'].fillna(df_moda['barrio'].mode()[0])
```

```python
print("- - -Imprimiendo tecnicas de manejo - - -")
print(f"Eliminando filas con Nulos:\n{df_borrado}")
print(f"\nImputando con Mediana:\n{df_mediana}")
print(f"\nimputando con Moda:\n{df_moda}")
```
